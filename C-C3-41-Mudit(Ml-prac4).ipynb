```python
import matplotlib.pyplot as plt
import seaborn as sns

numeric_cols = df.select_dtypes(
    include=["number"]
).columns  # gets only numeric columns

for col in numeric_cols:
  plt.figure(figsize=(7, 4))
  sns.histplot(df[col], kde=True)
  plt.title(f"Distribution of {col}")
  plt.xlabel(col)  # fixed typo from xlable
  plt.ylabel("Frequency")  # fixed typo from ylable
  plt.show()
```

```python
#outier analysis
for col in numeric_cols:
  plt.figure(figsize=(7, 4))
  sns.boxplot(x=df[col])  # creates a horizontal boxplot
  plt.title(f"Boxplot of {col}")
  plt.xlabel(col)
  plt.show()
```

```python
import pandas as pd

# detect outlier for every feature
# errors='ignore' prevents the KeyError if 'quantity' is missing or misspelled
features = df.drop(columns=["quantity"], errors="ignore").select_dtypes(
    include=["number"]
)
# calculate q1 and q3
Q1 = features.quantile(0.25)
Q3 = features.quantile(0.75)
IQR = Q3 - Q1
# calculate lower and upper limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
# detect outliers
outlier_mask = (features < lower_limit) | (features > upper_limit)
# count outliers for eachfeature
outlier_count = outlier_mask.sum()
# percentge of outliers
outlier_percentage = (outlier_count / len(features)) * 100
# summary table
outlier_summary = pd.DataFrame({
    "q1": Q1,
    "q3": Q3,
    "iqr": IQR,
    "lower": lower_limit,
    "upper": upper_limit,
    "outlier count": outlier_count,
    "outlier percentage": outlier_percentage,
})

print(outlier_summary)
```

```python
# 3. multivariate analysis
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(12, 8))
corr = df.select_dtypes(include=["number"]).corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix Heatmap")
plt.show()
```

```python

# pair plot

selected = ["alcohol", "volatile acidity", "sulphates", "citric acid", "quality"]

sns.pairplot(df[selected], hue="quality", palette="viridis")
plt.show()
```
